(f) Generate "vec_x" from a chi-square <br>

<font size = "3">

- Plot a histogram of each variable, with axis labels, and title
- Create a new variable "vec_y" by the formula <br>
$ y = e^x + 2x + 5$
- Plot "vec_x" against "vec_y" and label the axes

````python
# Write your own code
# import numpy as np
vec_x = np.random.chisquare(df=1, size=1000)
print (vec_x)
len(vec_x)
````

````python
plt.hist(vec_x)
plt.xlabel("vec_x") 
plt.ylabel("Frequency")  
plt.title("Histogram of vec_x") 
plt.show()

# Formula is given : y=ex+2x+5
# Create the variable: vec_y, vec_x against vec_y and label the axes.

vec_y = np.exp(vec_x) + 2*vec_x + 5
````

````python
# Formula is given : y=ex+2x+5
# np.exp(vec_x) →  eˣ, 2*vec_x →  2x, 5 →  +5
vec_y = np.exp(vec_x) + 2*vec_x + 5         
print(vec_y)
len(vec_y)
````

````python
# Created a variable vec_y using above code and here x axis is vec_y

plt.hist(vec_y)
plt.xlabel("vec_y")
plt.ylabel("Frequency")
plt.title("Histogram of vec_y")
plt.show()
````

````python
#Plot vec_x against vec_y and label the axes

plt.scatter(vec_x, vec_y)

plt.xlabel("vec_x")
plt.ylabel("vec_y")
plt.title("vec_x vs vec_y")

plt.show()
````

(g) Research online how to create an array with <br>
```size = 1000``` entries, with all entries equal to one.

<font size = "3">

- Create a new matrix, ```X_mat``` that appends a <br>
vector with only ones and "vec_x" from question (f) to form <br>
a matrix with two columns and 1000 rows
- Compute $(X_{mat}'X_{mat})^{-1}$

````python
# Write your own code

vec_ones = np.ones(1000)
print(vec_ones)
len(vec_ones) # checked length of the vector, it is 1000, as expected
vec_ones.shape
````

````python

X_mat = np.column_stack((vec_ones, vec_x))
print(X_mat)
X_mat.shape
````

````python
#X_mat.T → transpose of X_mat,  
# @ → matrix multiplication
# np.linalg.inv(...) → matrix inverse

result = np.linalg.inv(X_mat.T @ X_mat)
print(result) # here result is a 2x2 matrix, which is the inverse of the product of the transpose of X_mat and X_mat itself. This is often used in linear regression calculations to find the coefficients of the regression line.
````
