(c) Split a dataset into subsets

<font size = "3">

- Import the "features.csv" file as a pandas dataframe called "data"
- Create ``` list_status = [0,1]```
- Create ``` list_prob = [0.5, 0.5]```
- Create a new variable $n = 1000$ with the number of rows of the dataset
- Use 
```python
    np.random.choice(list_status,
                     size = n,
                     p = list_prob)
```
$\qquad$ to create a new column called "status" <br>
$\qquad$ which randomly assigns each row to "0" or "1" <br>
$\qquad$ with equal probability
- Use .query() to subset all observations with status == 0.<br>
store it in a new dataset called "data0"
- Use .query() to subset all observations with status == 1.<br>
store it in a new dataset called "data1"



````python
# Write your own code

import numpy as np

# Import the dataset
data = pd.read_csv("data_raw/features.csv")
print(data.head())

# Create the lists
list_status = [0, 1]
list_prob = [0.5, 0.5]
print(list_status)
print(list_prob)

# Number of rows in the dataset
n = len(data)
print(n)

# Create a new column called status
data["status"] = np.random.choice(
    list_status,
    size=n,
    p=list_prob
)
print(data.head())

# Create data0 containing status 0
data0 = data.query("status == 0")
print(data0.head())

# Create data1 containing status 1
data1 = data.query("status == 1")
print(data1.head())
````

(i) Plot by category

<font size = "3">

- Read "data_raw/wdi_urban.csv"<br>
This contains data from 1980 and 2020

- Create a list with the unique values of "year"
- Create a plot with a ...
    - Histogram of "prop_urbanpopulation" <br>
    for each year
    - Set the histogram option ``` alpha = 0.5 ``` for transparency
    - All the histograms should be overlapping
    - Create a legend with the list of years
    - Label the axes and the title

What do we learn from the graph?

HINT: See the example at the end of Lecture 10


````python
# Write your own code

import pandas as pd
import matplotlib.pyplot as plt

# Read the dataset
wdi_urban = pd.read_csv("data_raw/wdi_urban.csv")
print(wdi_urban.head())

# Create a list of unique years
list_years = pd.unique(wdi_urban["year"])
print(list_years)

# Create a histogram for each year
for year in list_years:
    df = wdi_urban.query("year == @year")
    plt.hist(x=df["prop_urbanpopulation"], alpha=0.5)


# Add labels, title, and legend
plt.xlabel("Proportion of Urban Population")
plt.ylabel("Frequency")
plt.title("Urban Population by Year")
plt.legend(labels=list_years, title="Year")

plt.show()




````

<font size = "3">

- Define $r = 0.04$
- Define two arrays $A$ and $T$ using "np.array()", assigning the following values

$ a = \begin{pmatrix} a_0 \\ a_1 \\ a_2 \end{pmatrix} =  \begin{pmatrix} 20 \\ 30 \\ 15 \end{pmatrix},

- Use element-by-element vector operations to compute a new array $M$ of length 3:

$ M = \begin{pmatrix} a_1 / (1+r)^{t_0}  \\ a_2 / (1+r)^{t_1} \\ a_3 / (1+r)^{t_2} \end{pmatrix} $

Note: I am using subscripts to denote the individual elements of $a$ and $t$.

````python
# Write your own code


import numpy as np

# Define r
r = 0.04

# Define arrays
a = np.array([20, 30, 15])
t = np.array([0, 1, 2])

# Compute M
M = a / (1 + r)**t

# Display M
print(M)

# Compute the sum
print(M.sum())


````
